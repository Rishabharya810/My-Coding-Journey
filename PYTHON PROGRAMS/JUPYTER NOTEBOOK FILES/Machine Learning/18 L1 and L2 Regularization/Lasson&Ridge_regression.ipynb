```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

dataset = pd.read_csv ("Melbourne_housing_FULL.csv")
dataset.head()

```

```python
dataset.nunique()
```

```python
dataset.shape
```

```python
cols_to_use = ['Suburb', 'Rooms', 'Type', 'Method', 'SellerG', 'Regionname', 'Propertycount', 
               'Distance', 'CouncilArea', 'Bedroom2', 'Bathroom', 'Car', 'Landsize', 'BuildingArea', 'Price']
dataset = dataset[cols_to_use]
```

```python
dataset.head()
```

```python
dataset.shape       
```

```python
dataset.isna().sum()
```

```python
cols_to_fill_zero = ['Propertycount', 'Distance', 'Bedroom2', 'Bathroom', 'Car']
dataset[cols_to_fill_zero] = dataset[cols_to_fill_zero].fillna(0)

dataset['Landsize'] = dataset['Landsize'].fillna(dataset.Landsize.mean())
dataset['BuildingArea'] = dataset['BuildingArea'].fillna(dataset.BuildingArea.mean())
```

```python
dataset.dropna(inplace=True)
dataset.shape
```

```python
dataset = pd.get_dummies(dataset, drop_first=True)
dataset.head()
```

```python
X = dataset.drop('Price', axis=1)
y = dataset['Price']
```

```python
from sklearn.model_selection import train_test_split
train_X, test_X, train_y, test_y = train_test_split(X, y, test_size=0.3, random_state=2)
```

```python
from sklearn.linear_model import LinearRegression
reg = LinearRegression().fit(train_X, train_y)
```

```python
reg.score(test_X, test_y)
```

```python
reg.score(train_X, train_y)
```

```python
from sklearn.linear_model import Lasso, Ridge
from sklearn.model_selection import GridSearchCV
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
import warnings
from sklearn.exceptions import ConvergenceWarning

lasso_alpha_grid = np.logspace(0, 4, 8)
ridge_alpha_grid = np.logspace(-4, 4, 12)

def tune_regularized_model(model, model_name, alpha_grid):
    pipeline = Pipeline([
        ("scaler", StandardScaler()),
        ("model", model),
    ])
    search = GridSearchCV(
        pipeline,
        param_grid={"model__alpha": alpha_grid},
        cv=5,
        scoring="r2",
        n_jobs=-1,
        return_train_score=True,
    )
    with warnings.catch_warnings():
        warnings.simplefilter("ignore", ConvergenceWarning)
        search.fit(train_X, train_y)
    results = (
        pd.DataFrame(search.cv_results_)[
            ["param_model__alpha", "mean_test_score", "std_test_score", "mean_train_score"]
        ]
        .sort_values("mean_test_score", ascending=False)
        .reset_index(drop=True)
    )
    print(f"{model_name} best alpha: {search.best_params_['model__alpha']}")
    print(f"{model_name} best CV R2: {search.best_score_:.4f}")
    # Score the refit pipeline directly; the scaler handles raw test_X internally.
    print(f"{model_name} train R2: {search.best_estimator_.score(train_X, train_y):.4f}")
    print(f"{model_name} test R2: {search.best_estimator_.score(test_X, test_y):.4f}")
    return search, results
```

```python
lasso_search, lasso_results = tune_regularized_model(Lasso(max_iter=50000, tol=1e-3, selection="random", random_state=42), "Lasso", lasso_alpha_grid)
lasso_results.head(10)
```

```python
ridge_search, ridge_results = tune_regularized_model(Ridge(), "Ridge", ridge_alpha_grid)
ridge_results.head(10)
```

```python
comparison = pd.DataFrame(
    {
        "Model": ["Lasso", "Ridge"],
        "Best alpha": [
            lasso_search.best_params_["model__alpha"],
            ridge_search.best_params_["model__alpha"],
        ],
        "Best CV R2": [lasso_search.best_score_, ridge_search.best_score_],
        "Train R2": [
            lasso_search.best_estimator_.score(train_X, train_y),
            ridge_search.best_estimator_.score(train_X, train_y),
        ],
        "Test R2": [
            lasso_search.best_estimator_.score(test_X, test_y),
            ridge_search.best_estimator_.score(test_X, test_y),
        ],
    }
).set_index("Model")

comparison
```
