```python
import numpy as np
import pandas as pd
import math
from sklearn.linear_model import LinearRegression
```

```python

# -----------------------------
# Load CSV
# -----------------------------
df = pd.read_csv("test_scores.csv")

x = df['math'].values
y = df['cs'].values
```

```python
# -----------------------------
# Gradient Descent Function
# -----------------------------
def gradient_descent(x, y, learning_rate=0.0002, max_iterations=1000000):
    m = 0
    b = 0
    n = len(x)
    prev_cost = float('inf')

    for i in range(max_iterations):
        y_pred = m * x + b

        # Mean Squared Error
        cost = (1/n) * np.sum((y - y_pred)**2)

        # Gradients
        md = -(2/n) * np.sum(x * (y - y_pred))
        bd = -(2/n) * np.sum(y - y_pred)

        # Update m and b
        m = m - learning_rate * md
        b = b - learning_rate * bd

        # Stop when cost stops improving
        if math.isclose(cost, prev_cost, rel_tol=1e-20):
            print(f"Stopped at iteration {i}")
            break

        prev_cost = cost

        # Print progress
        if i % 1000 == 0:
            print(f"Iteration {i}: m={m}, b={b}, cost={cost}")

    return m, b
```

```python
# -----------------------------
# Run Gradient Descent
# -----------------------------
m_gd, b_gd = gradient_descent(x, y)
print("\nGradient Descent Results:")
print("m =", m_gd)
print("b =", b_gd)
```

```python
# -----------------------------
# Sklearn Comparison
# -----------------------------
model = LinearRegression()
model.fit(df[['math']], df['cs'])

print("\nSklearn Results:")
print("m =", model.coef_[0])
print("b =", model.intercept_)
```
