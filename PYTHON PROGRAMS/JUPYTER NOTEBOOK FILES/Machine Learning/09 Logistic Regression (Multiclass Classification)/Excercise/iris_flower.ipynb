```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix, classification_report
%matplotlib inline
import matplotlib.pyplot as plt
import seaborn as sn
import pandas as pd

iris = load_iris()
```

```python
iris.data[0]
```

```python
iris.target[0:5]
```

```python
iris.feature_names, iris.target_names
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    iris.data,
    iris.target,
    test_size=0.2,
    random_state=42,
    stratify=iris.target,
)
```

```python
model = LogisticRegression(max_iter=200)
model.fit(X_train, y_train)
```

```python
model.score(X_test, y_test)
```

```python
sample_index = 0
predicted_flower = iris.target_names[model.predict([X_test[sample_index]])[0]].title()
predicted_flower
```

```python
actual_flower = iris.target_names[y_test[sample_index]].title()
actual_flower
```

```python
y_predicted = model.predict(X_test)

comparison = pd.DataFrame(
    X_test,
    columns=['Sepal Length', 'Sepal Width', 'Petal Length', 'Petal Width']
)
comparison['Predicted'] = [iris.target_names[value].title() for value in y_predicted]
comparison['Actual'] = [iris.target_names[value].title() for value in y_test]
comparison.head()
```

```python
cm = confusion_matrix(y_test, y_predicted)
cm
```

```python
labels = [name.title() for name in iris.target_names]

plt.figure(figsize=(8, 6))
sn.heatmap(cm, annot=True, fmt='d', xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted')
plt.ylabel('Truth')
plt.show()
```

```python
print(classification_report(y_test, y_predicted, target_names=iris.target_names))

mismatches = comparison[comparison['Predicted'] != comparison['Actual']]
print('Misclassified samples:', len(mismatches))
mismatches
```
